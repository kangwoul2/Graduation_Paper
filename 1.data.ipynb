```python
import yfinance as yf
import pandas as pd

etf_list = ["SPY", "QQQ", "DIA", "IWM", "EFA", "EEM", "XLK", "XLF", "XLE", "BTC-USD", "IAU"]

start_date = "2023-01-01"
end_date = "2024-01-20"

for etf in etf_list:
    data = yf.download(etf, start=start_date, end=end_date)
    
    data.rename(columns={
        "Open": "Opening Price",         
        "High": "Highest Price",        
        "Low": "Lowest Price",           
        "Close": "Closing Price",        
        "Adj Close": "Adjusted Closing Price",  
        "Volume": "Trading Volume"      
    }, inplace=True)
    
    filename = f"{etf}_ohlcv_data_final.csv"
    data.to_csv(filename)
    print(f"{filename} 저장 완료!")

print("모든 ETF 데이터 저장 완료!")
```
