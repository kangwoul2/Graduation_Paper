```python
import pandas as pd
import pandas_ta as ta
import glob

def add_technical_indicators(df):
    df = df.rename(columns={
        "Opening Price": "open",
        "Highest Price": "high",
        "Lowest Price": "low",
        "Closing Price": "close",
        "Adjusted Closing Price": "adj_close",
        "Trading Volume": "volume"
    })

    df["SMA_50"] = df.ta.sma(length=50, close=df["close"])
    df["EMA_20"] = df.ta.ema(length=20, close=df["close"])
    df["WMA_20"] = df.ta.wma(length=20, close=df["close"])
    
    macd = df.ta.macd(close=df["close"])
    df["MACD"] = macd["MACD_12_26_9"]
    df["MACD_signal"] = macd["MACDs_12_26_9"]
    df["MACD_hist"] = macd["MACDh_12_26_9"]
    
    df["TEMA_20"] = df.ta.tema(length=20, close=df["close"])

    df["RSI_14"] = df.ta.rsi(length=14, close=df["close"])
    df["ROC"] = df.ta.roc(length=10, close=df["close"])
    df["CCI_14"] = df.ta.cci(length=14, high=df["high"], low=df["low"], close=df["close"])
    df["WILLR_14"] = df.ta.willr(length=14, high=df["high"], low=df["low"], close=df["close"])

    df["ATR_14"] = df.ta.atr(length=14, high=df["high"], low=df["low"], close=df["close"])
    
    bbands = df.ta.bbands(length=20, close=df["close"])
    df["Upper_BB"] = bbands["BBU_20_2.0"]
    df["Middle_BB"] = bbands["BBM_20_2.0"]
    df["Lower_BB"] = bbands["BBL_20_2.0"]

    df["OBV"] = df.ta.obv(close=df["close"], volume=df["volume"])
    df["AD"] = df.ta.ad(high=df["high"], low=df["low"], close=df["close"], volume=df["volume"])
    df["Chaikin_AD"] = df.ta.adosc(high=df["high"], low=df["low"], close=df["close"], volume=df["volume"])

    return df

csv_files = glob.glob("*.csv")

for file in csv_files:
    df = pd.read_csv(file, parse_dates=["Date"])

    df_ta = add_technical_indicators(df)

    new_filename = file.replace(".csv", "_TA.csv")

    df_ta.to_csv(new_filename, index=False)
    print(f"{new_filename} 저장 완료!")
```
