```python
import pandas as pd
import glob
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

technical_indicators = [
    "sma_50", "ema_20", "wma_20", "macd", "macd_signal", "macd_hist", "tema_20",
    "rsi_14", "roc", "cci_14", "willr_14", "atr_14", "upper_bb", "middle_bb", "lower_bb",
    "obv", "ad", "chaikin_ad"
]

csv_files = glob.glob("3.Labeled_data/*_Labeled.csv")

for file in csv_files:
    df = pd.read_csv(file)

    df.columns = df.columns.str.strip().str.lower()

    if "date" in df.columns:
        df["date"] = pd.to_datetime(df["date"])  

    X = df[technical_indicators]
    y = df["label"]

    X = X.dropna()
    y = y[X.index]  

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)

    train_filename = file.replace("_Labeled.csv", "_Train.csv")
    test_filename = file.replace("_Labeled.csv", "_Test.csv")

    pd.DataFrame(X_train, columns=technical_indicators).assign(label=y_train.values).to_csv(train_filename, index=False)
    pd.DataFrame(X_test, columns=technical_indicators).assign(label=y_test.values).to_csv(test_filename, index=False)

    print(f"{train_filename}, {test_filename} 저장 완료!")
```
